**6회-1번) 소방서별 출동 소요시간 분석**

소방서는 신고를 접수한 후 현장으로 출동한다, 이때 신고 후 출동까지 걸린 시간을 분석하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 각 사건에 대한 '출동소요시간'을 구하여 컬럼을 추가하시오.
> 출동소요시간 = (출동시간 - 신고시간)
- 2) 1)의 결과를 사용하여 소방서(소방서명)별 '출동소요시간'의 평균을 구하시오.
- 3) 평균 출동소요시간 중 가장 큰 값을 구해, 반올림하여 분 단위로 변환하고, 그 값을 반올림하여 정수로 출력한다.

```python
df = pd.read_csv(path1 + "big_6_1_1.csv")
# print(df.head(3))

# 1. 각 사건에 대한 '출동소요시간'을 구하여 컬럼을 추가하시오.
# 출동소요시간 = (출동시간 - 신고시간)
df['출동시간'] = df['출동시간'].astype('datetime64[ns]')
df['신고시간'] = df['신고시간'].astype('datetime64[ns]')
df['출동소요시간'] = df['출동시간'] - df['신고시간']
# print(df.head(3))

# 2. 1)의 결과를 사용하여 소방서(소방서명)별 '출동소요시간'의 평균을 구하시오.
mean = df.groupby('소방서명')['출동소요시간'].transform('mean')
# print(mean)

# 평균 출동소요시간 중 가장 큰 값을 구해, 반올림하여 분 단위로 변환하고, 그 값을 반올림하여 정수로 출력한다.
print(round(mean.max().total_seconds() / 60)) # 3
```

**6회-2번) 교사 1인당 학생 수**

초등학교에서는 학년별 학생 수와 교사 수를 관리하고 있다.
학교별로 교사 1명이 맡고 있는 평균 학생 수를 계산하여, 관리가 필요한 학교를 파악하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 학교별 전체_학생수 및 전체_교사수를 구하시오.
- 2) 교사 1인당 학생 수를 계산하시오.
> 교사당_학생수 =  전체 학생수 / 전체 교사수
- 3) 교사 1인당 학생 수가 가장 많은 학교를 찾고, 해당 학교의 전체 교사 수를 정수로 제출합니다.

```python
df = pd.read_csv(path1 + "big_6_1_2.csv")
# print(df.head(3))

# 1. 학교별 전체_학생수 및 전체_교사수를 구하시오.
df2 = df.groupby('학교명')[['학생수','교사수']].sum().reset_index()
df2.rename(columns={'학생수': '전체_학생수', '교사수': '전체_교사수'}, inplace=True)
# print(df2.head(3))

# 2. 교사 1인당 학생 수를 계산하시오.(교사당_학생수 = 전체 학생수 / 전체 교사수)
df2['교사당_학생수'] = df2['전체_학생수'] / df2['전체_교사수']
# print(df2.head(3))

# 3. 교사 1인당 학생 수가 가장 많은 학교를 찾고, 해당 학교의 전체 교사 수를 정수로 제출합니다.
max_school = df2[df2['교사당_학생수'] == df2['교사당_학생수'].max()]['학교명'].values[0]
print(int(df2[df2['학교명'] == max_school]['전체_교사수'].values[0])) # 6
```

**6회-3번) 연도별 총 범죄 건수의 월평균 비교**

범죄 데이터를 분석하여, 연도별로 범죄 발생량의 특징을 파악하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 각 연도별로 범죄유형별 발생 건수를 모두 합산하여, 연도별 총 범죄 건수를 계산한다.
- 2) 1)에서 구한 연도별 총 범죄 건수를 12(개월)로 나누어 월평균 범죄 건수를 구한다.
- 3) 2)의 결과를 사용해 월평균 범죄 건수가 가장 큰 연도를 찾고,
해당 연도의 월평균 총 범죄 건수를 반올림하여 정수로 출력한다.

```python
df = pd.read_csv(path1 + "big_6_1_3.csv")
# print(df.head(3))

# 1. 각 연도별로 범죄유형별 발생 건수를 모두 합산하여, 연도별 총 범죄 건수를 계산한다.
df['연도'] = df['월별'].str[:4]
df['총범죄건수'] = df.loc[:, '강력범죄':'경범죄'].sum(axis=1)
df2 = df.groupby('연도', as_index=False)['총범죄건수'].sum()
# print(df2.head(3))

# 2. 1)에서 구한 연도별 총 범죄 건수를 12(개월)로 나누어 월평균 범죄 건수를 구한다.
df2['월평균 범죄건수'] = df2['총범죄건수'] / len(df['월별'].str[5:].unique())
# print(df2.head())

# 3. 2)의 결과를 사용해 월평균 범죄 건수가 가장 큰 연도를 찾고, 
# 해당 연도의 월평균 총 범죄 건수를 반올림하여 정수로 출력한다.
year = df2[df2['월평균 범죄건수'] == df2['월평균 범죄건수'].max()]['연도'].values[0]
print(round(df2[df2['연도'] == year]['월평균 범죄건수'].values[0])) # 4329
```

**14-4) 시간대별 평균 출동 소요시간 비교**

소방서는 하루 동안 다양한 시간대에 신고를 접수하고 현장에 출동한다.시간대별로 평균 출동 소요시간을 분석하여, 대응 효율성을 진단하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 출동 소요시간은 (출동시간 - 신고시간)으로 계산한다.
- 2)  신고시간을 기준으로 다음과 같이 시간대를 분류한다:
  - 새벽: 00:00 ~ 05:59
  - 오전: 06:00 ~ 11:59
  - 오후: 12:00 ~ 17:59
  - 밤: 18:00 ~ 23:59
- 3) 각 시간대별 출동 소요시간의 평균을 구한다.
- 4) 이 중 출동 소요시간 평균이 가장 큰 시간대의 출동 소요시간 평균을 분 단위로 변환하고, 그 값을 반올림하여 정수값을 제출하시오.


```python
df = pd.read_csv(path2 + "fire_station02.csv")
# print(df.head(3))

# 1. 출동 소요시간은 (출동시간 - 신고시간)으로 계산한다.
df['신고시간'] = df['신고시간'].astype('datetime64[ns]')
df['출동시간'] = df['출동시간'].astype('datetime64[ns]')
df['출동소요시간'] = df['출동시간'] - df['신고시간']
# print(df.head(3))

# 2. 신고시간을 기준으로 다음과 같이 시간대를 분류한다:
# 새벽: 00:00 ~ 05:59 오전: 06:00 ~ 11:59
# 오후: 12:00 ~ 17:59 밤: 18:00 ~ 23:59
df.loc[(df['신고시간'].dt.hour >= 0) & (df['신고시간'].dt.hour < 6), '시간대'] = '새벽'
df.loc[(df['신고시간'].dt.hour >= 6) & (df['신고시간'].dt.hour < 12), '시간대'] = '오전'
df.loc[(df['신고시간'].dt.hour >= 12) & (df['신고시간'].dt.hour < 18), '시간대'] = '오후'
df.loc[(df['신고시간'].dt.hour >= 18) & (df['신고시간'].dt.hour < 24), '시간대'] = '밤'
# print(df[['신고시간', '시간대']])

# 3. 각 시간대별 출동 소요시간의 평균을 구한다.
mean_times = df.groupby('시간대')['출동소요시간'].transform('mean')

# 이 중 출동 소요시간 평균이 가장 큰 시간대의 출동 소요시간 평균을 분 단위로 변환하고,
# 그 값을 반올림하여 정수값을 제출하시오.
print(round(mean_times.max().total_seconds() / 60)) # 7
```
