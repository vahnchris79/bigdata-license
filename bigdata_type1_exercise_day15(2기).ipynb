**7회-1번) 결측치 처리 및 과목별 점수 표준화**

학생들의 수업 과목별 점수 데이터를 변환하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 'score' 컬럼에서 결측치가 발생한 행을 제거한 뒤, 가장 많은 학생이 수강한 과목(subject)을 찾으시오.
- 2) 해당 과목의 점수(score)를 표준화(standardization)하여, 표준화된 점수 중 가장 높은 값을 구하시오.
> 표준화 공식: (값 - 평균) / 표준편차
- 3) 구한 값을 반올림하여 소수점 아래 3자리까지 출력한다.

```python
df = pd.read_csv(path1 + "student_scores01.csv")
# print(df.head(3), df.shape, sep="\n") # (1000, 4)
# print(df.isna().sum().to_frame().T)

# 1. 'score' 컬럼에서 결측치가 발생한 행을 제거한 뒤, 가장 많은 학생이 수강한 과목(subject)을 찾으시오.
df2 = df[~df['score'].isna()].copy()
# print(df.isna().sum().to_frame().T)
subject = df2['subject'].mode().max()

# 2. 해당 과목의 점수(score)를 표준화(standardization)하여, 표준화된 점수 중 가장 높은 값을 구하시오.
# 표준화 공식: (값 - 평균) / 표준편차
df3 = df2[df2['subject'] == subject]
mean = df3['score'].mean()
std = df3['score'].std(ddof=1)
df3['stard'] = [(df.loc[i, 'score'] - mean) / std for i in df3.index]
result = df3['stard'].max()

# 3. 구한 값을 반올림하여 소수점 아래 3자리까지 출력한다.
print(round(result, 3)) # 1.713
```

**7회-3번) 이상치 개수 구하기**

주어진 데이터에서 특정 기준에 따라 이상치 개수를 구하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 'feature_2' 컬럼의 이상치 개수를 구하여 정수로 출력한다.
- 2) 이상치 기준은 다음과 같다.
> 이상치 < 평균 - IQR * 1.5, 이상치 > 평균 + IQR * 1.5

```python
df = pd.read_csv(path1 + "outlier_data.csv")
# print(df.head(3), df.shape, sep="\n") # (2000, 10)

# 1. 'feature_2' 컬럼의 이상치 개수를 구하여 정수로 출력한다.
# 2. 이상치 기준은 다음과 같다.
# 이상치 < 평균 - IQR * 1.5, 이상치 > 평균 + IQR * 1.5
mean = df['feature_2'].mean()
Q1, Q3 = df['feature_2'].quantile([0.25, 0.75])
lower = mean - ((Q3 - Q1) * 1.5)
upper = mean + ((Q3 - Q1) * 1.5)
result = int(len(df[(df['feature_2'] < lower) | (df['feature_2'] > upper)]['feature_2']))
print(result) # 224
```

**15-4) 이상치가 가장 많은 독립변수 찾기 및 이상치 개수 구하기**

주어진 데이터는 아파트 가격 관련 데이터를 포함하고 있다. 아파트 가격(Price)은 종속변수이며, 나머지 변수들은 독립변수이다.

다음 절차에 따라 문제를 해결하시오.
- 1) 아파트 가격을 결정하는 독립변수들을 대상으로 분석을 수행한다.
- 2) 이상치 기준은 다음과 같다.
> 이상치 < 평균 - IQR * 1.5, 이상치 > 평균 + IQR * 1.5
- 3) 모든 독립변수에 대해 이상치 개수를 구한다.
- 4) 이상치 개수가 가장 많은 독립변수를 찾아, 이상치 개수를 정수로 출력합니다.

```python
df = pd.read_csv(path1 + "apartment_prices.csv")
# print(df.head(3))

# 1. 아파트 가격을 결정하는 독립변수들을 대상으로 분석을 수행한다.
# 2. 이상치 기준은 다음과 같다.
# 이상치 < 평균 - IQR * 1.5, 이상치 > 평균 + IQR * 1.5
# 3. 모든 독립변수에 대해 이상치 개수를 구한다.
# 4. 이상치 개수가 가장 많은 독립변수를 찾아, 이상치 개수를 정수로 출력합니다.
cnt_lists = []
for col in df.columns[:-1]:
    mean = df[col].mean()
    Q1, Q3 = df[col].quantile([0.25, 0.75])
    lower = mean - ((Q3 - Q1) * 1.5)
    upper = mean + ((Q3 + Q1) * 1.5)
    cnt = len(df[(df[col] < lower)|(df[col] > upper)])
    cnt_lists.append({f'{col}': cnt})
```

**15-5) 상관관계 분석**

주어진 데이터에서 독립변수들 간의 상관관계를 분석하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 결측치가 포함된 모든 행을 제거합니다.
- 2) 1)의 결과를 사용해 모든 숫자형 변수 간 상관계수를 'spearman'으로 계산하시오.
- 3) 숫자형 변수들 중 상관관계가 가장 큰 두 변수를 찾으시오.
- 단, 같은 변수 간의 상관계수(자기 자신과의 상관계수)는 제외한다.
- 4) 해당 두 변수의 상관계수 값을 반올림하여 소수점 아래 2자리까지 출력한다.

```python
import numpy as np
df = pd.read_csv(path1 + "penguins01.csv")
# print(df.head(3))
# print(df.isna().sum().to_frame().T)

# 1. 결측치가 포함된 모든 행을 제거합니다.
df = df.dropna(how='any')
# print(df.isna().sum().to_frame().T)

# 2. 1)의 결과를 사용해 모든 숫자형 변수 간 상관계수를 'spearman'으로 계산하시오.
cols_obj = df.select_dtypes(include='number').columns
corr = df[cols_obj].corr(method='pearson')

# 3. 숫자형 변수들 중 상관관계가 가장 큰 두 변수를 찾으시오.
# 단, 같은 변수 간의 상관계수(자기 자신과의 상관계수)는 제외한다.
for i in range(len(corr)):
    corr.iat[i, i] = 0
result = round(corr.max()[2], 2)
print(result) # 0.87
```
