### Day6. 다중 선형 회귀

**tips 데이터셋**
- total_bill : 식사 총 금액
- tip : 손님이 남긴 팁 금액
- sex : 서빙 받은 손님의 성별(Male, Female)
- smoker : 손님 일행에 흡연자가 포함되어 있는지 여부(Yes, No)
- day : 방문한 요일 (Thur, Fri, Sat, Sun)
- time : 식사시간대 (Lunch, Dinner)
- size : 일행의 인원 수 (손님 수)
- 244개 데이터

```python
# 데이터 확인
import pandas as pd
pd.set_option('display.width', 120)
filename = "https://raw.githubusercontent.com/Soyoung-Yoon/bigdata/main/tips.csv"
df = pd.read_csv(filename)
print(df.head(3), df.shape, sep="\n")
```

다음과 같은 다중선형회귀 모형을 사용한 회귀모델을 만들고 결과를 확인합니다.
- tips.csv 데이터를 사용합니다.
- 모델 생성시 상수항(=절편)을 포함하도록 합니다.
- 종속변수 : total_bill
- 독립변수 : tip, size, time, smoker

```python
# 6-1) 위의 조건에 맞게 선형 회귀 모델을 생성하고 summary를 출력한다.
from statsmodels.api import OLS
formula = 'total_bill ~ tip + size + C(time) + C(smoker)'
model = OLS.from_formula(formula, df).fit()
print(model.summary())
```

```python
#6-2) size가 2증가하면 total_bill이 몇 증가하는가? 결과는 반올림하여 소수점아래 4자리까지 출력한다.
print(round(model.params['size'] * 2, 4))
```
