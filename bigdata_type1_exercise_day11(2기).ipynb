**3회-1번) 캘리포니어 집값**  
주택 데이터에서 housing_median_age의 분포 파악을 위해 사분위수 범위(IQR)를 확인하려한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 결측치를 포함하는 모든 행을 제거한 후, 처음부터 순서대로 70%를 추출한다.
- 2) 'housing_median_age' 컬럼의 사분위 범위(IQR, Interquartile Range)의 값을 구한다.
> IQR = 3사분위수 - 1사분위수
- 3) 사분위 범위를 반올림하여 정수로 출력한다.

```python
df = pd.read_csv(path1 + "housing03.csv")
# print(df.head(3), df.info(), sep="\n") # 20640

# 결측치를 포함하는 모든 행을 제거한 후, 처음부터 순서대로 70%를 추출
df = df.dropna(axis=0)
# df.info() # 20433
df = df.loc[:int(len(df) * 0.7), :].copy()
# print(df.shape) # (14169, 10)

# 'housing_median_age' 컬럼의 사분위 범위(IQR, Interquartile Range)의 값을 구한다.
# IQR = 3사분위수 - 1사분위수
Q1, Q3 = df['housing_median_age'].quantile([0.25, 0.75])
IQR = Q3 - Q1

# 사분위 범위를 반올림하여 정수로 출력한다.
print(round(IQR)) # 19
```

**3회 2번) 연도별 나라별 유병률 데이터**

연도별 국가별 유병률 데이터를 사용하여, 특정 연도에 건강 지표가 상대적으로 높은 국가를 파악하고자 한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 2002년 데이터에 대해, 국가 전체 유병률 평균값보다 큰 유병률을 가진 국가의 개수를 구하시오.
- 2) 결과는 정수형으로 제출한다.

```python
df = pd.read_csv(path1 + "worlddata.csv")
# print(df.head(3))

# 2002년 데이터에 대해, 국가 전체 유병률 평균값보다 큰 유병률을 가진 국가의 개수를 구하시오
# 결과는 정수형으로 제출한다.
cols = [col for col in df.columns if col != 'year']
s = df.iloc[:, 1:].mean()
result = df[df['year'] == 2000].set_index('year').mean() > s
print(int(result.sum())) # 99
```

**3회-3번) 타이타닉 데이터 결측치 비율 분석**

**11-4) 범죄 유형의 발생 빈도 분석**

특정 범죄 유형이 평균 이상으로 발생한 연도 수를 파악하는 것은 추세 분석에 유용하다.

다음 절차에 따라 문제를 해결하시오.
- 1) '구분'이 '발생건수'인 데이터에서 '범죄유형5'의 연도별 발생건수를 구한다.
- 2) 1)의 결과에서 '범죄유형5'의 전체 평균 발생건수보다 큰 연도의 비율을 구한다.
- 비율 계산 방법
> 비율 = (평균보다 큰 연도 개수) ÷ (전체 연도 개수)
- 3) 최종 결과는 반올림하여 소수점 3자리까지 실수형으로 출력한다.

```python
df = pd.read_csv(path2 + "crime_data02.csv")
# print(df.head(3))

# '구분'이 '발생건수'인 데이터에서 '범죄유형5'의 연도별 발생건수를 구한다.
df5 = df.loc[df['구분']=='발생건수', ['연도','범죄유형5']].groupby('연도')['범죄유형5'].sum().reset_index(name='발생건수')

# 1)의 결과에서 '범죄유형5'의 전체 평균 발생건수보다 큰 연도의 비율을 구한다.
# 비율 계산 방법: 비율 = (평균보다 큰 연도 개수) ÷ (전체 연도 개수)
rate = df5[df5['발생건수'] > df5['발생건수'].mean()]['연도'].count() / df['연도'].count()

# 최종 결과는 반올림하여 소수점 3자리까지 실수형으로 출력한다.
print(round(rate, 3)) # 0.214
```

**11-5) 발생 비율이 높은 범죄유형**

년별, 범죄유형별 발생건수 데이터를 사용해 자주 많이 발생한 유형을 분석한다.

다음 절차에 따라 문제를 해결하시오.
- 1) 각 범죄유형별 연도별 발생건수를 구한다.
- 2) 1)번의 결과를 사용해 각 범죄유형에 대해, 전체 연도 동안의 평균 발생건수를 계산한다.
- 3) 각 범죄유형에 대해 평균보다 큰 발생건수를 기록한 연도의 비율을 구한다.
- 4) 3)에서 구한 비율이 0.5 이상인 범죄유형의 개수를 구해 정수로 출력한다.

```python
df = pd.read_csv(path2 + "crime_data02.csv")
# print(df.head(3))

# 1. 각 범죄유형별 연도별 발생건수를 구한다.
df2 = df.loc[df['구분']=='발생건수', '연도':'범죄유형10'].T.reset_index()
df2.columns = df2.iloc[0, :]
df2 = df2[2:]

# 2. 1.의 결과를 사용해 각 범죄유형에 대해, 전체 연도 동안의 평균 발생건수를 계산한다.
df2.columns = ['유형','2014','2015','2016','2017','2018','2019','2020']
df2['평균'] = df2.loc[:, '2014':'2020'].mean(axis=1)

# 3. 각 범죄유형에 대해 평균보다 큰 발생건수를 기록한 연도의 비율을 구한다.
years = ['2014','2015','2016','2017','2018','2019','2020']
df2['rate'] = df2[years].gt(df2['평균'], axis=0).sum(axis=1) / len(years)

# 4. 3)에서 구한 비율이 0.5 이상인 범죄유형의 개수를 구해 정수로 출력한다.
print(int(df2[df2['rate'] >=0.5]['유형'].count())) # 6
```
